```python
from textblob import TextBlob

# Let's test a sample positive phrase
sample_text_1 = "I am having an amazing day! The weather is beautiful."
blob_1 = TextBlob(sample_text_1)
print(f"Text 1: '{sample_text_1}'")
print(f"AI Polarity Score: {blob_1.sentiment.polarity}\n")

# Let's test a sample negative phrase
sample_text_2 = "This is worst error ever. I am so frustrated and tired."
blob_2 = TextBlob(sample_text_2)
print(f"Text 2: '{sample_text_2}'")
print(f"AI Polarity Score: {blob_2.sentiment.polarity}")
```

```python
from textblob import TextBlob

print("=====================================================")
print("🤖 AI Chat Assistant Subsystem Active")
print("Instructions: Type your sentence below and press Enter.")
print("To turn off the chatbot, type 'exit'")
print("=====================================================\n")

while True:
    # Capture user input inside Colab
    user_message = input("You: ")

    # Break loop if user wants to close the session
    if user_message.lower().strip() == 'exit':
        print("AI: Session terminated. Have a wonderful day!")
        break

    # Skip empty responses
    if not user_message.strip():
        continue

    # Feed text to the AI engine
    model_analysis = TextBlob(user_message)
    score = model_analysis.sentiment.polarity

    # Heuristic Classification Logic
    if score > 0.15:
        mood = "POSITIVE 🟢"
        bot_response = "AI: I can feel the fantastic energy in your words! Keep shining."
    elif score < -0.15:
        mood = "NEGATIVE 🔴"
        bot_response = "AI: That sounds incredibly rough. Take a deep breath; I'm here to listen."
    else:
        mood = "NEUTRAL 🟡"
        bot_response = "AI: Fascinating perspective. Tell me more about that."

    # Visual Output Dashboard
    print(f"   ↳ [System Log | Mood: {mood} | Numerical Value: {score:.2f}]")
    print(f"{bot_response}\n")
```

```python
# Cell 2: The Upgraded AI Brain using a Transformer Model
from transformers import pipeline

print("📥 Loading Deep Learning Sentiment Model into memory...")
# This automatically downloads a pre-trained DistilBERT model optimized for human language
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print("✅ Model successfully loaded!\n")

print("=====================================================")
print("🤖 AI Chat Assistant Subsystem (v2.0 Deep Learning)")
print("Instructions: Type your sentence below and press Enter.")
print("To turn off the chatbot, type 'exit'")
print("=====================================================\n")

while True:
    user_message = input("You: ")

    if user_message.lower().strip() == 'exit':
        print("AI: Session terminated. Have a wonderful day!")
        break

    if not user_message.strip():
        continue

    # Pass the text directly to the Neural Network
    prediction = classifier(user_message)[0]
    label = prediction['label']       # Will strictly return 'POSITIVE' or 'NEGATIVE'
    confidence = prediction['score']   # Confidence rating from 0.0 to 1.0

    # Process text parameters based on Neural Net context tracking
    if label == "POSITIVE" and confidence > 0.6:
        mood = "POSITIVE 🟢"
        bot_response = "AI: I can feel the fantastic energy in your words! Keep shining."
    elif label == "NEGATIVE" and confidence > 0.6:
        mood = "NEGATIVE 🔴"
        bot_response = "AI: That sounds incredibly rough. Take a deep breath; I'm here to listen."
    else:
        mood = "NEUTRAL 🟡"
        bot_response = "AI: Fascinating perspective. Tell me more about that."

    # Visual Output Dashboard showing the AI's math
    print(f"   ↳ [System Log | Neural Label: {label} | Confidence: {confidence:.2f}]")
    print(f"{bot_response}\n")
```

```python
from transformers import pipeline

print("📥 Loading Deep Learning Sentiment Model into memory...")
# This downloads a highly accurate, pre-trained AI model
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print("✅ Model successfully loaded!\n")

print("=====================================================")
print("🤖 AI Chat Assistant (Single-Run Edition)")
print("=====================================================\n")

# 1. Ask for your input EXACTLY ONCE
user_message = input("Type your sentence here and press Enter: ")

if user_message.strip():
    # 2. Pass your text directly to the Neural Network
    prediction = classifier(user_message)[0]
    label = prediction['label']       # Outputs 'POSITIVE' or 'NEGATIVE'
    confidence = prediction['score']   # Confidence score from 0.0 to 1.0

    # 3. Choose the response based on the AI's math
    if label == "POSITIVE" and confidence > 0.5:
        mood = "POSITIVE 🟢"
        bot_response = "AI: I can feel the fantastic energy in your words! Keep shining."
    else:
        # Since the model output is binary, if it's not positive, it's negative
        mood = "NEGATIVE 🔴"
        bot_response = "AI: That sounds incredibly rough. Take a deep breath; I'm here to listen."

    # 4. Display the beautiful final dashboard
    print(f"\n   ↳ [System Log | Neural Label: {label} | Confidence: {confidence:.2f}]")
    print(f"{bot_response}\n")

print("=====================================================")
print("✅ Execution finished cleanly. Click 'Play' to test a new phrase!")
print("=====================================================")
```
